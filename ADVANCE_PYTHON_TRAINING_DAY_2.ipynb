```python
class Area:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def returnArea(self):
        return self.length * self.breadth


length = float(input("Enter length:"))
breadth = float(input("Enter breadth:"))

obj = Area(length, breadth)

print("Area of Rectangle =", obj.returnArea())
```

```python
from abc import ABC, abstractmethod
#from abc import *
class Test(ABC):
  @abstractmethod
  def m1(self):
    pass

  @abstractmethod
  def m2(self):
    pass

class MyPython(Test):
  def m1(self):
    print("m1 implementation")
  def m2(self):
    print("m2 implementation")

m=MyPython()
m.m1()
m.m2()
```

If the child class is unable to provide implementation of all methods of abstract class then we can declare the child class methods with @abstractmethod and complete the remaining method implementation in next created child classes

```python
from abc import ABC, abstractmethod
#from abc import *
class Test(ABC):
  @abstractmethod
  def m1(self):
    pass

class MyPython(Test):
  def m1(self):
    print("m1 implementation")

  @abstractmethod
  def m2(self):
    pass

class Child(MyPython):
  def m2(self):
    print("m2 implementation")

m=Child()
m.m1()
m.m2()


```

```python
# We can use seek() function to move cursor (file pointer)
# specified location.

data = "Learning python is very easy from Ashish Sir"

f = open("myfile.txt", "w")
f.write(data)
f.close()

with open("myfile.txt", "r+") as f:
    fileData = f.read()
    print(fileData)

    print("Current Cursor Position is ", f.tell())

    f.seek(15)

    print("Current Cursor Position is ", f.tell())

    f.write("-JAVA-")

    f.seek(0)
    fileData = f.read()
    print(fileData)
```

```python
# WAP to print number of lines, words, characters present in a file

import os, sys

fname = input("Enter file name : ")

if os.path.isfile(fname):
    print("File exist : ", fname)
    f = open(fname, 'r')
else:
    print("File is not exist : ", fname)
    sys.exit(0)

lcount = wcount = ccount = 0

for line in f:
    lcount = lcount + 1
    ccount = ccount + len(line)
    words = line.split()
    wcount = wcount + len(words)

print("Total Lines : ", lcount)
print("Total words : ", wcount)
print("Total characters : ", ccount)

f.close()
```
