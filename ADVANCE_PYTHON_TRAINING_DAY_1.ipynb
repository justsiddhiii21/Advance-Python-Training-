```python
class A:
  def showA(self):
    print("Class A function")

class B(A):
  def showB(self):
    print("Class B function")

class C(A):
  def showC(self):
    print("Class C function")

class D(B,C):
  def showD(self):
    print("Class D function")

obj=D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()
```
