```python
import numpy as np
students=np.array([
     [85, 90, 78],#Student 1
     [60, 55, 70],#Student 2
     [95, 92, 88],#Student 3
     [40, 50, 45],#Student 4
     [72, 68, 75]#Student 5

 ])
students
```

```python
avg_marks= np.mean(students, axis=1)
print("Average Marks:", avg_marks)
```

```python
avg_marks= np.mean(students, axis=0)
print("Average Marks:", avg_marks)
```

```python
topper= np.argmax(avg_marks)
print("Topper Students:", topper)
```

```python
failed_students= np.where(avg_marks<50)
print("Failed Students:", failed_students[0]+1)
```

```python
failed_students= np.where(avg_marks<70)
print("Failed Students:", failed_students[0]+1)
```

```python
failed_students= np.where(avg_marks<80)
print("Failed Students:", failed_students[0]+1)
```

```python
failed_students= np.where(avg_marks<90)
print("Failed Students:", failed_students[0]+1)
```

```python
failed_students= np.where(avg_marks<100)
print("Failed Students:", failed_students[0]+1)
```

```python
#performance report
for i, avg in enumerate(avg_marks):
  status= "pass"
  if avg<50:
    status= "fail"
    print(f"Student {i+1}: Average Marks = {avg:.2f}")
  print(f"Student {i+1}: Average Marks = {avg:.2f}, Status = {status}")
```

```python
months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
sales = [250,300,350,400,450,500,550,600,650,700,750,800]
plt.plot(months,sales)
plt.title('Sales Analysis')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
#Bar Chart
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales = [250, 300, 400, 350, 450, 500, 550, 600, 650, 700, 750, 800]
plt.bar(months, sales)
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

```python
plt.barh(months, sales,color='Black')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.title('Monthly Sales Report')
plt.show()
```

```python
#Scatter Plot(comparing Height vs Weight)
height = [150, 160, 170, 180]
weight = [50, 60, 65, 75]
plt.scatter(height, weight, color='blue')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.title('Height vs Weight')
plt.show()
```

```python
#Scatter Plot(comparing Height vs Weight)
height = [150, 160, 170, 180]
weight = [50, 60, 65, 75]
plt.scatter(height, weight, color='pink')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.title('Height vs Weight')
plt.grid(True)
plt.show()
```

```python
#pie chart(market share)
brands = ['Brand A', 'Brand B', 'Brand C', 'Brand D']
market_share = [30, 25, 20, 15]
plt.pie(market_share, labels=brands, autopct='%1.1f%%', startangle=90)
plt.title('Market Share of Brands')
plt.show()
```

```python
#Multiple Plot in one figure
x= [1,2,3,4]
y1= [1,4,9,16]
y2= [1,2,3,4]
plt.plot(x, y1, label='Square')
plt.plot(x, y2, label='Linear')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Multiple Plots in One Figure')
plt.legend()
plt.title('Multiple Lines')
plt.show()
```

```python
#Subplot
fig,axs = plt.subplots(2,2)
axs[0,0].plot(x,y1)
axs[0,0].set_title('Square')

axs[0,1].bar(months,sales)
axs[0,1].set_title('sales')

axs[1,0].scatter(height,weight)
axs[1,0].set_title('scatter')

axs[1,1].pie(market_share,labels=brands,autopct='%1.1f%%',startangle=90)
axs[1,1].set_title('Pie')

plt.tight_layout()
plt.show()

```
